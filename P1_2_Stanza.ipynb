# Sesión 1 - Parte 2. Stanza

Stanza es una potente API de Procesamiento del Lenguaje Natural (PLN) desarrollada por Stanford. Permite realizar diversas tareas de PLN mediante la creación de "Pipelines" que procesan texto de manera secuencial.

El objetivo de esta práctica es crear un pipeline de Stanza en español y aplicar diferentes procesadores a un texto de ejemplo para entender cómo funciona cada etapa del pipeline.

Comenzaremos instalando la librería Stanza y descargando el modelo en español. Tened en cuenta que se pueden descargar modelos en distintos idiomas

```python
!pip3 install stanza
import stanza
stanza.download('es')
```

## Apartado 1.1 - Tokenización (Resuelto)

Vamos a crear nuestro pipeline paso a paso.

1.   Definiremos un texto de ejemplo en español.
2.   Crearemos un pipeline de Stanza, incluyendo paso a paso diferentes funciones de PLN.

En este apartado crearemos un pipeline de Stanza que incluya únicamente el proceso de "tokenize". Esto nos permitirá tokenizar el texto, es decir, dividirlo en unidades más pequeñas (tokens).

Vamos a mostrar cada token (word) de cada frase (sentence) presente en el texto.

```python
#Probamos el tokenizer en español
text = """Hugo come manzanas en la cocina de Telefónica.

Sofía juega al fútbol con Emma y Cristina con una pelota roja en Central Park.

El padre de Marina tiene 56 años.

La Tierra gira alrededor del Sol.

Júpiter es el planeta más grande del Sistema Solar"""

pipelineStanza = stanza.Pipeline(lang='es', processors='tokenize')
stanzaDoc = pipelineStanza(text)

print("Número de frases:" + str(len(stanzaDoc.sentences)))
for sentence in stanzaDoc.sentences:
  print(sentence)
```

```python
#Obtenemos todos los tokens por frase

for sentence in stanzaDoc.sentences:
  print("Frase:", sentence.text)
  for word in sentence.words:
    print(word.text)
  print('------------------------')
```

## Apartado 1.2 - Análisis Morfológico


En este apartado, ampliaremos nuestro pipeline para incluir el análisis morfológico.

Para ello añadiremos los procesos "mwt", y "pos". Esto nos permitirá realizar un análisis morfológico, identificando la categoría gramatical (UPOS) de cada token (work) de cada frase (sentence) presente en el texto.

Vamos a mostrar cada token junto con su categoría gramatical.

```python
pipelineStanza = stanza.Pipeline(lang='es', processors='tokenize,mwt,pos')
stanzaDoc = pipelineStanza(text)

print("Análisis Morfológico (Token y Categoría Gramatical UPOS):\n")
for sentence in stanzaDoc.sentences:
    print(f"Frase: {sentence.text}")
    for word in sentence.words:
        print(f"  Token: {word.text}, UPOS: {word.upos}")
    print('------------------------')
```

## Apartado 1.3 - Lemmatización

En este apartado, ampliaremos nuestro pipeline para incluir la **lemmatización**. Que es el proceso de agrupar las diferentes formas flexionadas de una palabra de forma que puedan ser analizadas como un único ítem, identificado por el lema del vocabulario, o forma base.

Esta es útil para reducir la dimensionalidad de los datos y analizar el significado fundamental de las palabras.

Para ello añadiremos el proceso "lemma". Esto nos permitirá realizar la lemmatización, obteniendo la forma base o diccionario de cada palabra (LEMMA) de cada token (work) de cada frase (sentence) presente en el texto.


Vamos a mostrar cada token junto con su lema entre paréntesis y su categoría gramatical.

```python
pipelineStanza = stanza.Pipeline(lang='es', processors='tokenize,mwt,pos,lemma')
stanzaDoc = pipelineStanza(text)

print("Numero de frases: " + str(len(stanzaDoc.sentences)))
print("Lematización (Token, LEMMA, Categoría Gramatical UPOS):\n")
for sentence in stanzaDoc.sentences:
    print(f"Frase: {sentence.text}")
    for word in sentence.words:
        print(f"  Token: {word.text}, LEMMA: {word.lemma}, UPOS: {word.upos}")
    print('------------------------')
```

## Apartado 1.4 - Análisis de Dependencias

En este apartado, ampliaremos nuestro pipeline para incluir un análisis de dependencias.

Para ello añadiremos el proceso "depparse". Esto nos permitirá realizar un análisis de dependencias, identificando las relaciones gramaticales entre las palabras de cada frase presente en el texto. Gracias a esto obtendremos la raíz principal (verbo principal) y su sujeto para cada frase, y opcionalmente, sus complementos directos.


Vamos a mostrar cada verbo principal junto con su sujeto y complementos directos para cada frase presente en el texto

```python
pipelineStanza = stanza.Pipeline(lang='es', processors='tokenize,mwt,pos,lemma,depparse')
stanzaDoc = pipelineStanza(text)

print("Numero de frases: " + str(len(stanzaDoc.sentences)))

for sentence in stanzaDoc.sentences:
    frase = ""
    root = ""
    subject = ""
    object = ""
    for word in sentence.words:
      frase += word.text + "(" + word.deprel + ")" + " "
      if word.deprel == "root":
        root = word.text
      if word.deprel == "nsubj":
        subject = word.text
      if word.deprel == "obj":
        object = word.text

    print(f"Frase: {frase}")
    print(f"  Raíz: {root}, Sujeto: {subject}, Objeto: {object}")
    print('------------------------')
```

## Apartado 1.5 - Reconocimiento de Entidades Nombradas (NER)

En este apartado, ampliaremos nuestro pipeline para incluir el reconocimiento de entidades nombradas (NER).

Para ello añadiremos el proceso "ner". Esto nos permitirá realizar un reconocimiento de entidades nombradas, identificando y clasificando las entidades presenes en el texto, como personas, organizaciones, lugares, etc.

Las entidades que puede detectar son:

| Etiqueta | Tipo | Ejemplos ilustrativos |
|---|---|---|
| `PER` | Personas | Hugo, Sofía, Gabriel García Márquez |
| `LOC` | Lugares | Madrid, España, Central Park |
| `ORG` | Organizaciones | Telefónica, ONU, Universidad de Murcia |
| `MISC` | Otras entidades nombradas | Nombres de acontecimientos u otras entidades que no encajan en las anteriores |

Vamos a mostrar las entidades detectadas en cada frase y su tipo.

```python
pipelineStanza = stanza.Pipeline(lang='es', processors='tokenize,mwt,pos,lemma,depparse,ner')
stanzaDoc = pipelineStanza(text)

print("Numero de frases: " + str(len(stanzaDoc.sentences)))

for sentence in stanzaDoc.sentences:
    print(f"Frase: {sentence.text}")
    if sentence.ents:
        for ent in sentence.ents:
            print(f"  Entidad: {ent.text}, Tipo: {ent.type}")
    else:
        print("  No se detectaron entidades.")
    print('------------------------')
```

### Mini resumen:

| Procesador | Qué aporta | Dónde mirar |
|---|---|---|
| `tokenize` | Oraciones y tokens | `doc.sentences`, `sent.tokens` |
| `mwt` | Expansión de tokens multipalabra | `token.words` |
| `pos` | Categorías y rasgos morfológicos | `word.upos` |
| `lemma` | Lema | `word.lemma` |
| `depparse` | Cabeza sintáctica y relación | `word.head`, `word.deprel` |
| `ner` | Entidades y sus tipos | `doc.ents`, `sent.ents` |
